## Что должно остаться после семинара

Ключевые связи:

$$L\text{-гладкость} \Longleftrightarrow \text{ограничение максимальной кривизны},$$

$$\text{выпуклость} \Longrightarrow \text{касательная — глобальная нижняя оценка},$$

$$\mu I\preceq\nabla^2 f(x)\preceq LI \Longrightarrow \kappa=\frac{L}{\mu}.$$

Эти величины будут напрямую определять скорость градиентного спуска на следующей лекции.
